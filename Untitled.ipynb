# Data Cleaning

```python
# convert upper to lower case

df['text']=df['text'].apply(lambda x:x.lower())
```

```python
# conver tokoneize
from nltk.tokenize import word_tokenize

df['transformate_text']=df['text'].apply(word_tokenize)
```

```python
# removing special character
df['transformate_text']=df['transformate_text'].apply(lambda x: [word for word in x if word .isalnum()])
```

```python
# stop  word
from nltk.corpus import stopwords
stop_word=set(stopwords.words('english'))
def remove_stop_word(text):
    new_text=[]
    for word in text:
        if word not in stop_word:
            new_text.append(word)
    return new_text
```

```python
df['transformate_text']=df['transformate_text'].apply(remove_stop_word)
```

```python
# Remove stemming 
from nltk.stem import PorterStemmer
ps=PorterStemmer()
df['transformate_text']=df['transformate_text'].apply(lambda x:[ps.stem(word) for word in x])
```

```python
# join list 

df['transformate_text']=df['transformate_text'].apply(lambda x:' '.join(x))
```

# model building

```python
from sklearn.feature_extraction.text import CountVectorizer,TfidfVectorizer
```

```python
cv=CountVectorizer()
tfidf=TfidfVectorizer(max_features=3000)
```

```python
x=tfidf.fit_transform(df['transformate_text']).toarray()
```

```python
y=df['target']
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
from sklearn.naive_bayes import GaussianNB,MultinomialNB,BernoulliNB
from sklearn.ensemble import RandomForestClassifier,GradientBoostingClassifier
from sklearn.metrics import accuracy_score,precision_score
```

```python
gnb=GaussianNB()
mnb=MultinomialNB()
bnb=BernoulliNB()
```

```python
gnb.fit(x_train,y_train)
y_pred=gnb.predict(x_test)
print(accuracy_score(y_test,y_pred))
print(precision_score(y_test,y_pred))
```

```python
mnb.fit(x_train,y_train)
y_pred1=mnb.predict(x_test)
print(accuracy_score(y_test,y_pred1))
print(precision_score(y_test,y_pred1))
```

```python
bnb.fit(x_train,y_train)
y_pred2=bnb.predict(x_test)
print(accuracy_score(y_test,y_pred2))
print(precision_score(y_test,y_pred2))
```
